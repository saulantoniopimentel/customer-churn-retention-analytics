```python
# Crear copia para Machine Learning
ml_df = df.copy()

# Convertir la variable objetivo a 0/1
ml_df["Churn"] = ml_df["Churn"].map({
    "No": 0,
    "Yes": 1
})

# El customerID no aporta información predictiva
ml_df = ml_df.drop(columns=["customerID"])

print("Filas:", ml_df.shape[0])
print("Columnas:", ml_df.shape[1])

ml_df.head()
```

```python
# Crear dataset limpio para Machine Learning
ml_df = df.copy()

# Eliminar variables creadas durante el análisis exploratorio
ml_df = ml_df.drop(
    columns=["customerID", "tenure_group", "high_risk_profile"]
)

# Convertir Churn a variable binaria
ml_df["Churn"] = ml_df["Churn"].map({
    "No": 0,
    "Yes": 1
})

print("Filas:", ml_df.shape[0])
print("Columnas:", ml_df.shape[1])
```

```python
# Variable objetivo
y = ml_df["Churn"]

# Variables predictoras
X = ml_df.drop(columns=["Churn"])

print("X:", X.shape)
print("y:", y.shape)
```

```python
# Variables categóricas
categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

# Variables numéricas
numeric_features = X.select_dtypes(
    exclude=["object"]
).columns.tolist()

print("Variables categóricas:")
print(categorical_features)

print("\nVariables numéricas:")
print(numeric_features)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "num",
            "passthrough",
            numeric_features
        )
    ]
)

print("Preprocesador creado correctamente.")
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)

print("\nDistribución de Churn en entrenamiento:")
print(y_train.value_counts(normalize=True))

print("\nDistribución de Churn en prueba:")
print(y_test.value_counts(normalize=True))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            LogisticRegression(
                max_iter=1000,
                random_state=42
            )
        )
    ]
)

print("Modelo Logistic Regression creado correctamente.")
```

```python
logistic_model.fit(X_train, y_train)

print("Modelo entrenado correctamente.")
```

```python
y_pred_logistic = logistic_model.predict(X_test)

print("Predicciones generadas:", len(y_pred_logistic))
```

```python
logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            LogisticRegression(
                max_iter=3000,
                random_state=42
            )
        )
    ]
)

logistic_model.fit(X_train, y_train)

y_pred_logistic = logistic_model.predict(X_test)

print("Modelo entrenado correctamente.")
print("Predicciones generadas:", len(y_pred_logistic))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Métricas
accuracy_logistic = accuracy_score(y_test, y_pred_logistic)
precision_logistic = precision_score(y_test, y_pred_logistic)
recall_logistic = recall_score(y_test, y_pred_logistic)
f1_logistic = f1_score(y_test, y_pred_logistic)

# Probabilidades para ROC-AUC
y_prob_logistic = logistic_model.predict_proba(X_test)[:, 1]
roc_auc_logistic = roc_auc_score(y_test, y_prob_logistic)

print("=== Logistic Regression ===")
print(f"Accuracy : {accuracy_logistic:.4f}")
print(f"Precision: {precision_logistic:.4f}")
print(f"Recall   : {recall_logistic:.4f}")
print(f"F1-Score : {f1_logistic:.4f}")
print(f"ROC-AUC  : {roc_auc_logistic:.4f}")

print("\n=== Classification Report ===")
print(classification_report(y_test, y_pred_logistic))
```

```python
cm_logistic = confusion_matrix(y_test, y_pred_logistic)

print("Matriz de confusión:")
print(cm_logistic)
```

```python
from sklearn.ensemble import RandomForestClassifier

random_forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=300,
                max_depth=12,
                min_samples_leaf=5,
                random_state=42,
                n_jobs=-1,
                class_weight="balanced"
            )
        )
    ]
)

random_forest_model.fit(X_train, y_train)

print("Random Forest entrenado correctamente.")
```

```python
# Predicciones
y_pred_rf = random_forest_model.predict(X_test)

# Probabilidades
y_prob_rf = random_forest_model.predict_proba(X_test)[:, 1]

# Métricas
accuracy_rf = accuracy_score(y_test, y_pred_rf)
precision_rf = precision_score(y_test, y_pred_rf)
recall_rf = recall_score(y_test, y_pred_rf)
f1_rf = f1_score(y_test, y_pred_rf)
roc_auc_rf = roc_auc_score(y_test, y_prob_rf)

print("=== Random Forest ===")
print(f"Accuracy : {accuracy_rf:.4f}")
print(f"Precision: {precision_rf:.4f}")
print(f"Recall   : {recall_rf:.4f}")
print(f"F1-Score : {f1_rf:.4f}")
print(f"ROC-AUC  : {roc_auc_rf:.4f}")

print("\n=== Classification Report ===")
print(classification_report(y_test, y_pred_rf))
```

```python
cm_rf = confusion_matrix(y_test, y_pred_rf)

print("Matriz de confusión:")
print(cm_rf)
```

```python
# Obtener probabilidad de churn
churn_probability = random_forest_model.predict_proba(X_test)[:, 1]

# Convertir a score de 0 a 100
churn_risk_score = churn_probability * 100

print("Churn Risk Score:")
print(f"Mínimo: {churn_risk_score.min():.2f}")
print(f"Máximo: {churn_risk_score.max():.2f}")
print(f"Promedio: {churn_risk_score.mean():.2f}")
```

```python
import numpy as np

risk_category = np.select(
    [
        churn_risk_score >= 70,
        churn_risk_score >= 40
    ],
    [
        "Alto",
        "Medio"
    ],
    default="Bajo"
)

print(pd.Series(risk_category).value_counts())
```

```python
# Índices de los clientes utilizados en el conjunto de prueba
test_indices = X_test.index

# Crear tabla base con información original
risk_table = df.loc[test_indices].copy()

print("Clientes en tabla de riesgo:", risk_table.shape[0])
```

```python
# Agregar resultados del modelo
risk_table["churn_probability"] = churn_probability
risk_table["risk_score"] = churn_risk_score
risk_table["risk_category"] = risk_category

# Convertir probabilidad a porcentaje
risk_table["churn_probability_pct"] = (
    risk_table["churn_probability"] * 100
)

print(risk_table.shape)
```

```python
risk_table[
    [
        "customerID",
        "Contract",
        "tenure",
        "MonthlyCharges",
        "InternetService",
        "PaymentMethod",
        "Churn",
        "churn_probability_pct",
        "risk_score",
        "risk_category"
    ]
].sort_values(
    "risk_score",
    ascending=False
).head(10)
```

```python
powerbi_df = risk_table[
    [
        "customerID",
        "gender",
        "SeniorCitizen",
        "Partner",
        "Dependents",
        "tenure",
        "tenure_group",
        "PhoneService",
        "MultipleLines",
        "InternetService",
        "OnlineSecurity",
        "OnlineBackup",
        "DeviceProtection",
        "TechSupport",
        "StreamingTV",
        "StreamingMovies",
        "Contract",
        "PaperlessBilling",
        "PaymentMethod",
        "MonthlyCharges",
        "TotalCharges",
        "Churn",
        "churn_probability",
        "churn_probability_pct",
        "risk_score",
        "risk_category",
        "high_risk_profile"
    ]
].copy()

print("Filas:", powerbi_df.shape[0])
print("Columnas:", powerbi_df.shape[1])
```

```python
powerbi_df = powerbi_df.sort_values(
    "risk_score",
    ascending=False
).reset_index(drop=True)

powerbi_df.head(10)
```

```python
# Entrenar el modelo final con todos los clientes
final_rf_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", RandomForestClassifier(
            n_estimators=300,
            max_depth=12,
            min_samples_leaf=5,
            random_state=42,
            n_jobs=-1,
            class_weight="balanced"
        ))
    ]
)

final_rf_model.fit(X, y)

print("Modelo final entrenado correctamente.")
```

```python
# Predecir probabilidad de churn para todos los clientes
all_churn_probability = final_rf_model.predict_proba(X)[:, 1]

# Convertir a porcentaje / score de riesgo
all_risk_score = all_churn_probability * 100

# Crear categorías de riesgo
all_risk_category = np.select(
    [
        all_risk_score >= 70,
        all_risk_score >= 40
    ],
    [
        "Alto",
        "Medio"
    ],
    default="Bajo"
)

print("Clientes evaluados:", len(all_risk_score))
print("Riesgo promedio:", round(all_risk_score.mean(), 2))
```

```python
# Crear tabla final de riesgo para todos los clientes
risk_table_all = df.copy()

risk_table_all["churn_probability"] = all_churn_probability
risk_table_all["churn_probability_pct"] = all_churn_probability * 100
risk_table_all["risk_score"] = all_risk_score
risk_table_all["risk_category"] = all_risk_category

# Ordenar por mayor riesgo
risk_table_all = risk_table_all.sort_values(
    "risk_score",
    ascending=False
).reset_index(drop=True)

print("Filas:", risk_table_all.shape[0])
print("Columnas:", risk_table_all.shape[1])

risk_table_all.head(10)
```

```python
# Distribución de clientes por categoría de riesgo
risk_distribution = (
    risk_table_all["risk_category"]
    .value_counts()
    .reindex(["Bajo", "Medio", "Alto"])
    .reset_index()
)

risk_distribution.columns = ["Categoría de riesgo", "Clientes"]

risk_distribution["Porcentaje"] = (
    risk_distribution["Clientes"] /
    len(risk_table_all) * 100
)

risk_distribution
```
